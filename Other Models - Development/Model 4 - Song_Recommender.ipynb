```python
from datasets import load_dataset
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer

# Function to load the dataset, filter English songs, and display its summary
def load_song_dataset():
    try:
        # Load the dataset
        dataset = load_dataset("Nicolas-BZRD/English_French_Songs_Lyrics_Translation_Original")
        df = pd.DataFrame(dataset['train'])

        # Filter only English songs
        df = df[df['language'] == 'en']

        # Display summary
        print(f"Dataset Shape: {df.shape}")
        print("Column Names:", df.columns.tolist())

        # Display 10 sample records
        print("\nSample Records:")
        print(df.head(10))

        print("\nDataset loaded successfully.")
        return df
    except Exception as e:
        print(f"Error loading dataset: {e}")
        return pd.DataFrame()  # Return an empty DataFrame in case of failure

# Generate embeddings for song lyrics
def generate_embeddings(df, model, batch_size=32):
    if 'embedding' not in df.columns:
        print("Generating embeddings...")

        # Process in batches
        embeddings = []
        for i in range(0, len(df), batch_size):
            batch = df['original_version'].iloc[i:i+batch_size].tolist()
            batch_embeddings = model.encode(batch, convert_to_tensor=True, show_progress_bar=True)
            embeddings.extend(batch_embeddings)

        df['embedding'] = [e.tolist() for e in embeddings]
    else:
        print("Embeddings already exist.")
    return df

# Function to recommend similar songs
def recommend_songs(song_title, df):
    if song_title not in df['title'].values:
        return "Song not found in the dataset."

    song_idx = df[df['title'] == song_title].index[0]
    query_embedding = np.array(df.loc[song_idx, 'embedding'])

    # Prepare matrix of all embeddings
    embedding_matrix = np.vstack(df['embedding'].to_numpy())

    # Compute cosine similarities
    similarities = cosine_similarity([query_embedding], embedding_matrix).flatten()

    # Exclude the selected song
    similarities[song_idx] = -1  # Exclude the selected song by setting its score to -1

    # Get top recommendations
    top_indices = np.argsort(similarities)[::-1][:5]
    recommended_songs = df.iloc[top_indices][['artist_name', 'album_name', 'title', 'original_version']].to_dict('records')
    return recommended_songs

if __name__ == "__main__":
    # Load and preprocess dataset
    song_data_df = load_song_dataset()

    if not song_data_df.empty:
        # Load SentenceTransformer model
        model = SentenceTransformer('all-MiniLM-L6-v2')

        # Generate embeddings for the dataset
        song_data_df = generate_embeddings(song_data_df, model)

        # Example: Recommend similar songs based on user input
        user_input_song = "Let It Be"  # Example song title
        recommendations = recommend_songs(user_input_song, song_data_df)

        print("\nRecommended Songs:")
        for idx, rec in enumerate(recommendations):
            print(f"{idx+1}. {rec['artist_name']} - {rec['title']} ({rec['album_name']})")
            print(f"   Lyrics: {rec['original_version'][:100]}...\n")
```
